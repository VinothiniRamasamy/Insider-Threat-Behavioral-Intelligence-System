```python
datasets = [logon, device, email, file, http]

for df in datasets:
    df['hour'] = df['date'].dt.hour
    df['day'] = df['date'].dt.day
    df['month'] = df['date'].dt.month
    df['year'] = df['date'].dt

print("Time Features Created Successfully")
```

```python
insiders = pd.read_csv(
    "/kaggle/input/datasets/andrihjonior/cert-insider-threat-dataset-r4-2/answers/insiders.csv"
)

insiders.head()
```

```python
print(insiders.shape)
print(insiders.columns)
```

```python
insiders.info()
```

```python
insiders['start'] = pd.to_datetime(
    insiders['start'],
    format='mixed',
    errors='coerce'
)

insiders['end'] = pd.to_datetime(
    insiders['end'],
    format='mixed',
    errors='coerce'
)

print(insiders[['start', 'end']].isnull().sum())
```

```python
insiders[insiders['start'].isna() | insiders['end'].isna()]
```

```python
insiders = insiders.dropna(subset=['start', 'end'])

print("Remaining rows:", len(insiders))
```

```python
insider_users = insiders['user'].unique()

print("Total Insider Users:", len(insider_users))
print(insider_users[:10])
```

```python
logon['label'] = logon['user'].isin(insider_users).astype(int)
device['label'] = device['user'].isin(insider_users).astype(int)
email['label'] = email['user'].isin(insider_users).astype(int)
file['label'] = file['user'].isin(insider_users).astype(int)
http['label'] = http['user'].isin(insider_users).astype(int)

print("Labels created successfully")
```

```python
print("Logon")
print(logon['label'].value_counts())

print("\nDevice")
print(device['label'].value_counts())

print("\nEmail")
print(email['label'].value_counts())

print("\nFile")
print(file['label'].value_counts())

print("\nHTTP")
print(http['label'].value_counts())
```
